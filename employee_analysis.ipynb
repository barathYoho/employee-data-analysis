```python
# Create original dataset backup
df_raw = df.copy()

# Create working dataset
df_clean = df.copy()

print("Original dataset backup created successfully!")
```

```python
# Check missing values before cleaning
print(df_clean.isnull().sum())
```

```python
# Count duplicates before removal
duplicates_before = df_clean.duplicated().sum()

# Remove duplicates
df_clean = df_clean.drop_duplicates()

print("Duplicates removed:", duplicates_before)
print("Remaining duplicates:", df_clean.duplicated().sum())
```

```python
for col in df_clean.select_dtypes(include="object").columns:

    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Text standardisation completed!")
```

```python
# Detect outliers in numerical columns

numeric_cols = df_clean.select_dtypes(include="number").columns

for col in numeric_cols:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower) |
        (df_clean[col] > upper)
    ]

    print(f"\nColumn: {col}")
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Outliers detected:", len(outliers))
```

```python
# Before cleaning
before = {
    "Rows": len(df_raw),
    "Missing Values": df_raw.isnull().sum().sum(),
    "Duplicates": df_raw.duplicated().sum()
}

# After cleaning
after = {
    "Rows": len(df_clean),
    "Missing Values": df_clean.isnull().sum().sum(),
    "Duplicates": df_clean.duplicated().sum()
}

# Create comparison table
summary = pd.DataFrame({
    "Before Cleaning": before,
    "After Cleaning": after
})

display(summary)
```

```python
duplicates_before = df_clean.duplicated().sum()

df_clean.drop_duplicates(inplace=True)

print("Duplicates removed:", duplicates_before)
print("Remaining duplicates:", df_clean.duplicated().sum())
```

```python
for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Unnecessary spaces removed successfully!")
```

```python
# Standardise text columns
for col in df_clean.select_dtypes(include=["str", "object"]).columns:
    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Text standardisation completed without the warning!")
```

```python
numeric_cols = df_clean.select_dtypes(include="number").columns

for col in numeric_cols:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower_limit) |
        (df_clean[col] > upper_limit)
    ]

    print("\nColumn:", col)
    print("Outliers detected:", len(outliers))
```

```python
from pathlib import Path

files = list(Path.cwd().rglob("clean_data.csv"))

for file in files:
    print("Found:", file)
```

```python
df = pd.read_csv(files[0])

print("Dataset loaded successfully!")
display(df.head())
```

```python
print("Columns:", df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# Salary statistical analysis

salary_stats = df["salary"].agg([
    "mean",
    "median",
    "min",
    "max",
    "std"
])

print("EMPLOYEE SALARY STATISTICS")
display(salary_stats)
```

```python
department_analysis = df.groupby("department").agg(
    Employee_Count=("id", "count"),
    Average_Salary=("salary", "mean"),
    Maximum_Salary=("salary", "max"),
    Minimum_Salary=("salary", "min")
).round(2)

print("DEPARTMENT-WISE ANALYSIS")
display(department_analysis)
```

```python
city_analysis = df.groupby("city").agg(
    Employee_Count=("id", "count"),
    Average_Salary=("salary", "mean"),
    Maximum_Salary=("salary", "max")
).round(2)

print("CITY-WISE ANALYSIS")
display(city_analysis)
```

```python
top_employees = df.nlargest(3, "salary")

print("TOP 3 HIGHEST-PAID EMPLOYEES")
display(top_employees[["name", "department", "salary", "city"]])
```

```python
numeric_data = df[["salary"]]

print("Available business numerical variables:")
print(numeric_data.columns.tolist())

print("\nCorrelation matrix:")
display(numeric_data.corr())
```

```python
# 1. Which department has the highest average salary?
print("1. Highest average salary department:")
display(department_analysis.nlargest(1, "Average_Salary"))

# 2. Which city has the most employees?
print("\n2. City with most employees:")
display(city_analysis.nlargest(1, "Employee_Count"))

# 3. What is the average employee salary?
print("\n3. Average employee salary:")
print("Rs.", df["salary"].mean())

# 4. Who are the three highest-paid employees?
print("\n4. Top three highest-paid employees:")
display(top_employees[["name", "salary"]])

# 5. Which department has the most employees?
print("\n5. Department with most employees:")
display(department_analysis.nlargest(1, "Employee_Count"))
```

```python
import os

os.makedirs("reports", exist_ok=True)

salary_stats.to_csv("reports/salary_statistics.csv")
department_analysis.to_csv("reports/department_analysis.csv")
city_analysis.to_csv("reports/city_analysis.csv")
top_employees.to_csv("reports/top_employees.csv", index=False)

print("EDA results saved successfully!")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs("reports/figures", exist_ok=True)

df = pd.read_csv("reports/clean_data.csv")

sns.set_theme(style="whitegrid")

print("Visualization setup completed!")
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="department",
    y="salary",
    estimator="mean"
)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (Rs.)")

plt.tight_layout()
plt.savefig("reports/figures/bar_chart.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(df["salary"], bins=5, edgecolor="black")

plt.title("Employee Salary Distribution")
plt.xlabel("Salary (Rs.)")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.savefig("reports/figures/histogram.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="id",
    y="salary",
    hue="department",
    s=100
)

plt.title("Employee ID vs Salary")
plt.xlabel("Employee ID")
plt.ylabel("Salary (Rs.)")

plt.tight_layout()
plt.savefig("reports/figures/scatter_plot.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sorted_df = df.sort_values("id")

plt.plot(
    sorted_df["id"],
    sorted_df["salary"],
    marker="o"
)

plt.title("Employee Salary Comparison")
plt.xlabel("Employee ID")
plt.ylabel("Salary (Rs.)")

plt.tight_layout()
plt.savefig("reports/figures/line_chart.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="department",
    y="salary"
)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary (Rs.)")

plt.tight_layout()
plt.savefig("reports/figures/box_plot.png", dpi=300)
plt.show()
```
